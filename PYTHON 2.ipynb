Q2 Sales Filtering Analysis

Filter the dataset to display:

Orders where Sales > 3500

Orders where Region = "East" and Sales > 2500

Explain what insights you observe. 


```python
# 1. Filter orders where Sales > 3500
sales_gt_3500 = df[df['Sales'] > 3500]
print("--- Orders where Sales > 3500 (First 5 rows) ---")
print(sales_gt_3500.head())
print(f"Total orders with Sales > 3500: {len(sales_gt_3500)}")

# 2. Filter orders where Region = 'East' and Sales > 2500
east_sales_2500 = df[(df['Region'] == 'East') & (df['Sales'] > 2500)]
print("\n--- Orders in East Region with Sales > 2500 (First 5 rows) ---")
print(east_sales_2500.head())
print(f"Total rows: {len(east_sales_2500)}")
```

Observed Insights:
High-Value Concentration: There are 104 orders (out of 230 total rows) where sales exceed 3500, indicating that the business generates a significant portion of its total volume from high-ticket transactions.
Regional Contribution: The East region alone accounts for 48 orders with sales greater than 2500, showing that it is a particularly strong market for higher-priced product sales.

Q3 Customer Purchase Analysis

Identify the top 5 customers based on total sales.

Hint: Use GroupBy with aggregation. 


```python
# Group by customer and calculate total sales, then get the top 5
top_customers = df.groupby('Customer')['Sales'].sum().nlargest(5)

print("--- Top 5 Customers by Total Sales ---")
print(top_customers)
```

Q4 Regional Sales Performance

Calculate the total sales generated by each region.

Then identify:

Which region generated the highest revenue?

```python
# Group by region and calculate total sales
region_sales = df.groupby('Region')['Sales'].sum()

print("--- Total Sales by Region ---")
print(region_sales)

# Identify the region with the highest revenue
highest_region = region_sales.idxmax()
print(f"\nRegion generating the highest revenue: {highest_region} ({region_sales[highest_region]})")
```

Q5 Product Performance Analysis

Calculate the total quantity sold for each product.

Which product appears to be the most popular?

```python
# Calculate total quantity sold for each product
prod_quantity = df.groupby('Product')['Quantity'].sum()

print("--- Total Quantity Sold by Product ---")
print(prod_quantity)

# Identify the most popular product
most_popular = prod_quantity.idxmax()
print(f"\nMost Popular Product: {most_popular} ({prod_quantity[most_popular]} units)")
```

Q6 Multi-Level Grouping

Group the dataset by: Region and Category

Calculate the total sales for each combination.

What insights can you derive from this analysis? 


```python
# Group the dataset by Region and Category and calculate total sales
reg_cat_sales = df.groupby(['Region', 'Category'])['Sales'].sum()

print("--- Total Sales by Region and Category ---")
print(reg_cat_sales)
```

Insights Derived from this Analysis:
Electronics Dominance: Electronics consistently generates the highest total sales across all regions, with a particularly massive contribution in the East region.
Office Supplies Lag: The Office category consistently records the lowest sales volume across every single region, indicating lower market demand or smaller ticket sizes compared to Accessories and Electronics.
Regional Preferences: While Electronics leads overall, categories like Accessories show strong, steady sales performance across the South and West regions, highlighting diverse regional buying behaviors

Q7 Pivot Table Creation

Create a pivot table showing:

Rows → Region

Columns → Product

Values → Total Sales

Which region sells the most laptops?

```python

pivot_laptop = df.pivot_table(index='Region', columns='Product', values='Sales', aggfunc='sum')

print("--- Pivot Table: Region vs Product Sales ---")
print(pivot_laptop)

top_laptop_region = pivot_laptop['Laptop'].idxmax()
print(f"\nRegion that sells the most Laptops: {top_laptop_region} ({pivot_laptop.loc[top_laptop_region, 'Laptop']})")
```

Q8 Profitability Analysis

Calculate the average profit for each product category.

Which category is most profitable? 


```python

avg_profit_cat = df.groupby('Category')['Profit'].mean()

print("--- Average Profit by Product Category ---")
print(avg_profit_cat)

most_profitable = avg_profit_cat.idxmax()
print(f"\nMost Profitable Category: {most_profitable} ({avg_profit_cat[most_profitable]:.2f})")
```

Q10 End-to-End Data Wrangling Task

Perform the following workflow:

Filter the dataset where Sales > 3000

Group data by Region and Product

Calculate total sales

Reshape the result into a pivot table

Explain the business insights obtained from this analysis.

```python
# 1. Filter the dataset where Sales > 3000
filtered_df = df[df['Sales'] > 3000]

# 2 & 3. Group data by Region and Product and calculate total sales
grouped_q10 = filtered_df.groupby(['Region', 'Product'])['Sales'].sum().reset_index()

# 4. Reshape the result into a pivot table
pivot_q10 = grouped_q10.pivot_table(index='Region', columns='Product', values='Sales', fill_value=0)

print("--- End-to-End Pivot Table (Sales > 3000) ---")
print(pivot_q10)
```

Business Insights Obtained:
Concentrated High-Value Drivers: High-value transactions (Sales > 3000) are heavily concentrated in specific product categories per region. For example, Laptops are the dominant high-ticket sales driver in the East region, whereas Keyboards lead high-value sales in the West region.
Targeted Inventory & Marketing: By filtering out lower-tier orders, management can pinpoint which specific products generate premium revenue in each geographic market, enabling more effective regional inventory stocking and tailored marketing campaigns.
